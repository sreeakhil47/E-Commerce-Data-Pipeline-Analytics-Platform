```python

import pandas as pd
from pathlib import Path

# Folder containing your uploaded CSV files
DATA_DIR = Path("/content")
OUTPUT_DIR = DATA_DIR / "typed_data"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

# All 8 datasets: none are optional
file_names = {
    "Customers": "customers.csv",
    "Products": "products.csv",
    "Sellers": "sellers.csv",
    "Orders": "orders.csv",
    "Order_Items": "order_items.csv",
    "Payments": "payments.csv",
    "Reviews": "reviews.csv",
    "Analytics": "analytics_cleaned.csv"
}

# Identifier columns: keep as text
id_columns = [
    "customer_id", "product_id", "seller_id",
    "order_id", "order_item_id", "payment_id",
    "review_id", "event_id", "transaction_id"
]

# Date and timestamp columns
date_columns = [
    "order_date", "registration_date", "payment_date",
    "review_date", "event_date", "event_timestamp",
    "timestamp", "created_at", "updated_at"
]

# Numeric columns
numeric_columns = [
    "price", "unit_price", "subtotal", "total_amount",
    "amount", "revenue", "discount", "tax",
    "shipping_cost", "rating", "quantity",
    "stock_quantity", "inventory", "age"
]

integer_columns = [
    "quantity", "stock_quantity", "inventory", "age", "rating"
]

# Verify all required files exist before processing
missing_files = [
    filename for filename in file_names.values()
    if not (DATA_DIR / filename).is_file()
]

if missing_files:
    raise FileNotFoundError(
        "Upload or rename these required files first:\n"
        + "\n".join(missing_files)
    )

typed_datasets = {}
type_reports = []

# Process every dataset
for dataset_name, filename in file_names.items():

    print(f"\nProcessing {dataset_name}...")

    # Read all columns as strings initially to preserve IDs
    df = pd.read_csv(
        DATA_DIR / filename,
        dtype="string",
        keep_default_na=False
    )

    # Standardize column names
    df.columns = (
        df.columns.str.strip()
        .str.lower()
        .str.replace(r"[^a-z0-9]+", "_", regex=True)
        .str.strip("_")
    )

    # Standardize blank cells to missing values
    df = df.replace(r"^\s*$", pd.NA, regex=True)

    # Trim spaces in text fields
    for col in df.select_dtypes(
        include=["object", "string"]
    ).columns:
        df[col] = df[col].str.strip()

    # Convert ID columns to string
    for col in id_columns:
        if col in df.columns:
            df[col] = df[col].astype("string")

    # Convert date and timestamp columns
    for col in date_columns:
        if col in df.columns:
            cleaned = df[col].replace(
                ["N/A", "NA", "null", "None", ""], pd.NA
            )
            df[col] = pd.to_datetime(
                cleaned, errors="coerce"
            )

    # Convert numeric fields
    for col in numeric_columns:
        if col in df.columns:
            cleaned = df[col].replace(
                ["N/A", "NA", "null", "None", ""], pd.NA
            )
            numeric = pd.to_numeric(
                cleaned, errors="coerce"
            )

            if col in integer_columns:
                df[col] = numeric.astype("Int64")
            else:
                df[col] = numeric

    # Save the converted dataset
    output_path = OUTPUT_DIR / f"{dataset_name.lower()}_typed.csv"
    df.to_csv(output_path, index=False)

    typed_datasets[dataset_name] = df

    # Record resulting data types
    for col in df.columns:
        type_reports.append({
            "Dataset": dataset_name,
            "Column": col,
            "Data_Type": str(df[col].dtype)
        })

    print("Rows:", len(df))
    print("Columns:", len(df.columns))
    print("Saved:", output_path)
    print(df.dtypes.to_string())

# Save a single report for all 8 datasets
type_report = pd.DataFrame(type_reports)
type_report.to_csv(
    OUTPUT_DIR / "all_datasets_data_types.csv",
    index=False
)

print("\nAll 8 datasets processed successfully.")
print("Data type report saved.")
display(type_report)
```

# **12. Checking The Relationships Between Datasets**

```python

import pandas as pd
from pathlib import Path

# 1. Load all CSV datasets from Colab
DATA_DIR = Path("/content")

datasets = {}

for file in DATA_DIR.glob("*.csv"):
    name = file.stem.lower().replace(" ", "_")

    # Skip output files created during earlier cleaning
    if name.endswith("_cleaned"):
        continue

    # Standardize dataset names
    name = name.replace("-", "_")
    datasets[name] = pd.read_csv(file, dtype="string")

# 2. Standardize column names and values
for name, df in datasets.items():
    df.columns = (
        df.columns.str.strip()
        .str.lower()
        .str.replace(r"[^a-z0-9]+", "_", regex=True)
        .str.strip("_")
    )

    for col in df.select_dtypes(
        include=["object", "string"]
    ).columns:
        df[col] = df[col].str.strip()

print("Datasets found:", list(datasets.keys()))


# 3. Find a dataset using possible filename variations
def get_dataset(name):
    for key, df in datasets.items():
        if key == name or key.startswith(name + "_"):
            return df
    return None


# 4. Define primary keys to check
primary_keys = {
    "customers": ["customer_id"],
    "products": ["product_id"],
    "sellers": ["seller_id"],
    "orders": ["order_id"],
    "order_items": ["order_item_id"],
    "payments": ["payment_id"],
    "reviews": ["review_id"],
}

print("\n========== PRIMARY KEY CHECKS ==========")

for table, columns in primary_keys.items():
    df = get_dataset(table)

    if df is None:
        print(f"{table}: Dataset not found")
        continue

    for col in columns:
        if col not in df.columns:
            print(f"{table}: Column '{col}' not found")
            continue

        ids = df[col].dropna()
        ids = ids[~ids.str.lower().isin(["", "n/a", "nan", "null"])]

        duplicates = ids[ids.duplicated(keep=False)]

        print(
            f"{table}.{col}: "
            f"{len(duplicates)} rows with duplicate IDs; "
            f"{df[col].isna().sum()} missing IDs"
        )


# 5. Define foreign key relationships
relationships = [
    ("orders", "customer_id", "customers", "customer_id"),
    ("order_items", "order_id", "orders", "order_id"),
    ("order_items", "product_id", "products", "product_id"),
    ("order_items", "seller_id", "sellers", "seller_id"),
    ("payments", "order_id", "orders", "order_id"),
    ("reviews", "order_id", "orders", "order_id"),
    ("reviews", "customer_id", "customers", "customer_id"),
    ("reviews", "product_id", "products", "product_id"),
    ("analytics", "order_id", "orders", "order_id"),
    ("analytics", "customer_id", "customers", "customer_id"),
    ("analytics", "product_id", "products", "product_id"),
    ("analytics", "seller_id", "sellers", "seller_id"),
    ("products", "seller_id", "sellers", "seller_id"),
]

print("\n========== FOREIGN KEY CHECKS ==========")

results = []

missing_tokens = ["", "n/a", "nan", "null", "none", "<na>"]

for child_table, child_col, parent_table, parent_col in relationships:
    child = get_dataset(child_table)
    parent = get_dataset(parent_table)

    # Skip relationships whose datasets or columns do not exist
    if child is None or parent is None:
        continue

    if child_col not in child.columns or parent_col not in parent.columns:
        continue

    child_ids = child[child_col].astype("string").str.strip()
    parent_ids = parent[parent_col].astype("string").str.strip()

    # Exclude missing/placeholder references from orphan checks
    valid_child = child_ids[
        child_ids.notna()
        & ~child_ids.str.lower().isin(missing_tokens)
    ]

    valid_parent = parent_ids[
        parent_ids.notna()
        & ~parent_ids.str.lower().isin(missing_tokens)
    ]

    parent_set = set(valid_parent)

    orphan_mask = ~valid_child.isin(parent_set)
    orphan_count = int(orphan_mask.sum())
    missing_count = int(
        (child_ids.isna()
         | child_ids.str.lower().isin(missing_tokens)).sum()
    )

    results.append({
        "relationship": (
            f"{child_table}.{child_col} -> "
            f"{parent_table}.{parent_col}"
        ),
        "orphan_references": orphan_count,
        "missing_references": missing_count,
        "status": "CHECK REQUIRED" if orphan_count else "PASS"
    })

    print(
        f"{child_table}.{child_col} -> "
        f"{parent_table}.{parent_col}: "
        f"{orphan_count} orphan references, "
        f"{missing_count} missing references"
    )

# 6. Save the relationship validation report
if results:
    report = pd.DataFrame(results)
    report.to_csv(
        "/content/relationship_validation_report.csv",
        index=False
    )

    print("\n========== SUMMARY ==========")
    display(report)

    print("Report saved as relationship_validation_report.csv")
else:
    print(
        "No relationships were checked. "
        "Verify dataset filenames and column names."
    )
```
